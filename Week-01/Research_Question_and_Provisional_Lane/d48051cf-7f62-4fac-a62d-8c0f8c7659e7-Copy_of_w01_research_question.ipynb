## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

My work will be able to provide **directional insights** and **decision-support** by identifying patterns in customer behavior that are **observed** to correlate with churn. It can help prioritize which customers are at higher risk and thus inform targeted retention strategies. This is a model based on historical data, meaning it reflects past trends and correlations. What it *cannot* do is establish **causal proof** — it cannot definitively say *why* a customer churns, only that certain characteristics or behaviors are associated with higher churn probability. It also cannot 'predict Google' or other external market shifts; its utility is within the scope of the available internal customer data. The recommendations will be **observational** and need validation through A/B testing or other controlled experiments before being fully deployed.
